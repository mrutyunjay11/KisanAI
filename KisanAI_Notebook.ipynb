## 📋 What This Project Does
KisanAI is a production-ready AI agent for Indian farmers combining all 5 days of the intensive:
- **Day 1** — Agent core: understands farmer queries in natural language
- **Day 2** — Live tool integrations: Open-Meteo Weather API, data.gov.in Mandi Price API, Gemini AI
- **Day 3** — Long-term farmer memory: name, crop, location, last 5 questions
- **Day 4** — Safety guardrails: pesticide dose referrals, harmful query blocking
- **Day 5** — Production-ready: analytics dashboard, error fallbacks, observability logging

## 🛠️ APIs Used
| API | Purpose | Cost |
|-----|---------|------|
| Open-Meteo | Live weather + geocoding | Free (no key) |
| Google Search Grounding | Live crop mandi prices | Free tier (via Gemini) |
| Google Gemini 3.5 Flash | AI chat + intelligence | Free tier |

## 🏗️ Architecture
```
Farmer types question (English / Hindi / Gujarati)
            ↓
    [Day 1: Agent Core — Intent Detection]
            ↓
  ┌─────────────────────────────────────┐
  │        Day 2: LIVE TOOLS            │
  │  🌦️  Open-Meteo Weather API        │
  │  💰  Gemini Mandi Search Grounding │
  │  🏛️  Gemini generates schemes       │
  │  🌱  Gemini crop pest advice        │
  └─────────────────────────────────────┘
            ↓
    [Day 3: Farmer Memory — name, crop, location, 5 past Qs]
            ↓
    [Day 4: Safety Guardrails — block harmful requests]
            ↓
    [Gemini 3.5 Flash — warm personalized reply]
            ↓
    [Day 5: Analytics + Logging + Error Fallbacks]
```
